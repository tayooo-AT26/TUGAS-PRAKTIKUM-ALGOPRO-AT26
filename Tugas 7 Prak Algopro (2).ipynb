```python
%%writefile saving_project/interest.py
def compound_interest(principal, rate, time):   # Fungsi untuk menghitung bunga majemuk
    Future value = principal * (1 + rate) ** time   # Rumus: saldo awal x (1 + bunga) pangkat lama menabung

def simple_interest(principal, rate, time):    # Fungsi untuk menghitung bunga sederhana
    return principal * (1 + rate * time)    # Rumus: saldo awal x (1 + bunga x lama menabung)
```

```python
%%writefile saving_project/display.py
"""Modul display: hanya berisi fungsi untuk menampilkan hasil."""

def print_result(principal, rate, years, ci, si):
    """Tampilkan saldo awal, bunga tahunan, lama menabung,
    nilai akhir bunga majemuk (ci) dan bunga sederhana (si)."""
    print("\n===== HASIL PERHITUNGAN TABUNGAN =====")
    print(f"{'Saldo Awal':<18}: ¥{principal:,}")    # {principal:,} menambahkan pemisah ribuan, contoh 80000.0 -> 80,000.0

    # str() mengubah angka menjadi teks supaya bisa digabung dengan tanda +
    print(f"{'Bunga Tahunan':<18}: {rate * 100}%")  # rate dikali 100 agar 0.05 tampil sebagai 5.0%
    print(f"{'Lama Menabung':<18}: {years} tahun")
    print("-" * 37)
    
    # {ci:,.2f} artinya: pemisah ribuan dan 2 angka di belakang koma
    print(f"Nilai Akhir (Bunga Majemuk)  : ¥{ci:,.2f}")    
    print(f"Nilai Akhir (Bunga Sederhana): ¥{si:,.2f}")
    print("=" * 37)
```

```python
%%writefile saving_project/main.py
"""Program utama: Kalkulator Bunga Tabungan."""

# Mengambil fungsi dari file interest.py dan display.py
import interest
import display

# Meminta input dari pengguna
def main():
    print("=== Program Kalkulator Bunga Tabungan ===")
    principal = float(input("Masukkan saldo awal : "))
    rate = float(input("Masukkan bunga tahunan (misal 3% → 0.03): "))
    years = int(input("Masukkan lama menabung (tahun): "))

# Memanggil fungsi dari modul interest untuk menghitung nilai akhir
ci = interest.compound_interest(principal, rate, years)      # bunga majemuk
si = interest.simple_interest(principal, rate, years)        # bunga sederhana

display.print_result(principal, rate, years, ci, si)


if __name__ == "__main__":
    main()
```
