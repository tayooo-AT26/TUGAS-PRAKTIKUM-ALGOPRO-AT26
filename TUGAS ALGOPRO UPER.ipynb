```python
#NOMOR 5, MENJUAL DORAYAKI SEBANYAK 350 PCS DENGAN HARGA 5.000/PCS
#MENGHITUNG PENGHASILAN DARI TOKO MENGGUNAKAN OPERASI ARITMATIKA

harga = 5000
terjual = 350 
penghasilan = (harga * terjual)
print ('Jumlah penghasilan:', penghasilan)
```

```python
#NOMOR 6, KEUNTUNGAN TOKO DORAYAKI BULAN PERTAMA 
#MENGHITUNG KEUNTUNGAN DARI TOKO MENGGUNAKAN OPERASI ARITMATIKA

penghasilan = 1750000
modal = 1110000
keuntungan = penghasilan - modal 
print ('Total keuntungan dalam sebulan:', keuntungan)
```

```python
#NOMOR 7, BETULKAN KODE YANG SALAH

print ("Keuntungan total adalah" keuntungan) 
```
