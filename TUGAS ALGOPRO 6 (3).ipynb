```python
# A. VARIABEL GLOBAL
skor_nobita = 0                      # Menyimpan skor Nobitaa, awalnya 0

#B. FUNGSI UPDATE SKOR
def update_skor(nilai):              # Fungsi untuk mengubah skor
    global skor_nobita               # Memakai variabel skor_nobita yang ada di luar fungsi
    skor_nobita += nilai             # Tambah skor (jika nilai negatif, skor berkurang)

#C. FUNGSI AKTIVITAS NOBITA
# Fungsi belajar
def belajar(jam_belajar):                              # Menerima jumlah jam belajar
    if jam_belajar < 1:                                # Jika belajar kurang dari 1 jam
        print("Nobita hanya membuka buku sebentar")    # Tampilkan pesan
        return -50                                     # Kembalikan skor -50
    elif 1 <= jam_belajar < 3:                         # Jika belajar 1 sampai kurang dari 3 jam
        print("Nobita belajar secukupnya")             # Tampilkan pesan
        return 25                                      # Kembalikan skor +25
    else:                                              # Jika belajar 3 jam atau lebih
        print("Nobita belajar dengan rajin hari ini")  # Tampilkan pesan
        return 100                                     # Kembalikan skor +100

# Fungsi tidur
def tidur(jam_tidur):                                  # Menerima jumlah jam tidur
    if jam_tidur < 1:                                  # Jika tidur kurang dari 1 jam
        print("Nobita kurang tidur")                   # Tampilkan pesan
        return -10                                     # Kembalikan skor -10
    elif 1 <= jam_tidur <= 2:                          # Jika tidur 1 sampai 2 jam
        print("Waktu tidur Nobita seimbang")           # Tampilkan pesan
        return 20                                      # Kembalikan skor +20
    else:                                              # Jika tidur lebih dari 2 jam
        print("Nobita tidur terlalu lama")             # Tampilkan pesan
        return -20                                     # Kembalikan skor -20

# Fungsi main bersama Doraemon
def main_bersama_doraemon(jumlah_gadget, waktu_main):                # Menerima jumlah gadget dan waktu main
    if waktu_main <= 1 and jumlah_gadget <= 2:                       # Main maksimal 1 jam DAN gadget maksimal 2
        print("Nobita bermain secukupnya")                           # Tampilkan pesan
        return 15                                                    # Kembalikan skor +15
    elif waktu_main <= 2 and jumlah_gadget <= 3:                     # Main maksimal 2 jam DAN gadget maksimal 3
        print("Nobita masih bisa mengatur waktu dengan baik")        # Tampilkan pesan
        return 5                                                     # Kembalikan skor +5
    else:                                                            # Selain kondisi di atas
        print("Nobita kebanyakan bermain")                           # Tampilkan pesan
        return -30                                                   # Kembalikan skor -30

# Fungsi menyontek
def menyontek():                                       # Tidak butuh input apa pun
    print("Nobita ketahuan menyontek")                 # Tampilkan pesan
    return -50                                         # Kembalikan skor -50

#D. FUNGSI PENGENDALI AKTIVITAS
def jalankan_aktivitas(pilihan):                                      # Menerima pilihan menu dari user
    if pilihan == 1:                                                  # Jika pilih 1 (Belajar)
        jam = float(input("Berapa jam Nobita belajar hari ini? "))    # Input jam belajar
        skor = belajar(jam)                                           # Hitung skor belajar
        update_skor(skor)                                             # Masukkan skor ke total
    elif pilihan == 2:                                                # Jika pilih 2 (Tidur)
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))# Input jam tidur
        skor = tidur(jam)                                             # Hitung skor tidur
        update_skor(skor)                                             # Masukkan skor ke total
    elif pilihan == 3:                                                # Jika pilih 3 (Main)
        gadget = int(input("Berapa gadget yang digunakan Nobita? "))  # Input jumlah gadget
        waktu = float(input("Berapa jam Nobita bermain gadget? "))    # Input lama bermain
        skor = main_bersama_doraemon(gadget, waktu)                   # Hitung skor bermain
        update_skor(skor)                                             # Masukkan skor ke total
    elif pilihan == 4:                                                # Jika pilih 4 (Menyontek)
        skor = menyontek()                                            # Hitung skor menyontek
        update_skor(skor)                                             # Masukkan skor ke total
    else:                                                             # Jika pilihan selain 1-4
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")   # Tampilkan pesan error

#E. FUNGSI EVALUASI AKHIR
def valuasi_akhir():                   # Menentukan kategori berdasarkan total skor
    if skor_nobita >= 80:              # Skor 80 ke atas
        return "Siswa Teladan"         # Kategori terbaik
    elif 50 <= skor_nobita < 80:       # Skor 50 sampai kurang dari 80
        return "Cukup Rajin"           # Kategori menengah
    else:                              # Skor di bawah 50
        return "Pemalas"               # Kategori terendah

#F. MENU UTAMA PROGRAM
def main():                                                  # Fungsi utama program
    print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")       # Judul program
    print("1. Belajar")                                      # Tampilkan menu 1
    print("2. Tidur")                                        # Tampilkan menu 2
    print("3. Main bersama Doraemon")                        # Tampilkan menu 3
    print("4. Menyontek saat ujian")                         # Tampilkan menu 4

    pilihan = int(input("Pilih aktivitas (1-4): "))          # Input pilihan user (diubah ke angka)
    jalankan_aktivitas(pilihan)                              # Jalankan aktivitas sesuai pilihan

    print("\n--- HASIL EVALUASI ---")                        # Judul hasil
    print(f"Total Skor Saat Ini : {skor_nobita}")            # Tampilkan total skor
    print(f"Kategori Status     : {valuasi_akhir()}")        # Tampilkan kategori status
    
#JALANKAN PROGRAM UTAMA
if __name__ == "__main__":                                   # Jalankan hanya jika file ini dieksekusi langsung
    main()                                                   # Panggil fungsi

def main():
    print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main bersama Doraemon")
    print("4. Menyontek saat ujian")

    pilihan = 1                                   # Pilihan otomatis: Belajar
    print("Pilih aktivitas (1-4): ", pilihan)     # Tampilkan pilihan
    jam = 4                                       # Jam belajar otomatis: 4
    print("Berapa jam Nobita belajar hari ini? ", jam)
    update_skor(belajar(jam))                     # Hitung dan tambahkan skor

    print("\n--- HASIL EVALUASI ---")
    print(f"Total Skor Saat Ini : {skor_nobita}")
    print(f"Kategori Status     : {valuasi_akhir()}")
```
