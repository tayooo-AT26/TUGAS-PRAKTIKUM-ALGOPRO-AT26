```python
# Input
usia_pengemudi = int(input("Usia pengemudi: "))
jenis_kendaraan = input("Jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = int(input("Nilai kendaraan (Rp): "))
area_penggunaan = input("Area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Memiliki sistem pengaman? (ya/tidak): ").lower()

# 1. Usia pengemudi (operator boolean)
#    Terlalu muda (<20) atau terlalu tua (>60) -> risiko tinggi
risiko_usia = "tinggi" if usia_pengemudi < 20 or usia_pengemudi > 60 else "rendah"
    
# 2. Tentukan tingkat risiko awal berdasarkan jenis kendaraan (if-else)
if jenis_kendaraan == "motor":
    risiko = "sedang"
else: 
    risiko = "rendah"

# 3. Dasar perhitungan premi (if-elif-else)
if nilai_kendaraan > 500000000:
    risiko_nilai = "tinggi"
    persentase_premi = 0.10
elif 100000000 <= nilai_kendaraan < 500000000:
    risiko_nilai = "sedang"
    persentase_premi = 0.07
else:
    risiko_nilai = "rendah"
    persentase_premi = 0.05


# 4. Area penggunaan (operator boolean)
#    Jika di perkotaan, risiko meningkat satu level
if area_penggunaan == "perkotaan" and risiko == "rendah": 
    risiko = "sedang"
elif area_penggunaan == "perkotaan" and risiko == "sedang":
    risiko = "tinggi"

# 5. Sistem pengaman (if-elif-else)
#    Jika tidak punya sistem keamanan, risiko meningkat satu level
if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"
    else:
        risiko = "tinggi"

# 6. Hitung premi
premi = persentase_premi * nilai_kendaraan

# 7. Pesan Doraemon (conditional expression)
pesan = ("Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
         if risiko == "tinggi"
         else "Risiko kendaraan masih aman menurut Doraemon")

# Output
print("\n===== HASIL ANALISIS =====")
print("1. Kategori risiko:", risiko)
print("2. Premi asuransi yang perlu dibayarkan: Rp", int(premi))
print("3. Pesan Doraemon:", pesan)
```
