```python
#A. MENCATAT PENGELUARAN HARIAN NOBITA

hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input(f"Masukkan pengeluaran hari ke-{hari}: "))
    saldo -= pengeluaran
    if saldo <= 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
    else:
        print(f"Pengeluaran hari ke-{hari}: {pengeluaran}")
        print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")
        hari += 1
```

```python
#B. MENGANALISIS PENGELUARAN MENGGUAN NOBITA 

pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    minggu = i + 1
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{minggu}: Nobita tidak jajan, dilewati analisisnya.")
        continue
    print(f"Minggu ke-{minggu}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis mingguan selesai.")
```

```python
#C. MENYUSUN LAPORAN BULANAN NOBITA

for bulan in range(1, 4):
    print(f"Bulan ke-{bulan}")
    total_bulanan = 0  # Menggunakan nama variabel total_bulanan, bukan sum
    
    for minggu in range(1, 5):
        # Menggunakan int() untuk konversi input dengan benar
        pengeluaran = int(input(f"Pengeluaran minggu ke-{minggu}: "))
        print(f"Pengeluaran minggu ke-{minggu}: {pengeluaran}")
        
        if pengeluaran > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            break
            
        total_bulanan += pengeluaran
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")
        
    print()
```

```python
#D. EVALUASI AKHIR TAHUN KEUANGAN NOBITA 

saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print(f"=== Bulan ke-{bulan} ===")
    minggu = 1
    
    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")
        
        # Cetak kembali kalimat dan data inputan user
        print(f"Apakah minggu ke-{minggu} libur? (y/n): {libur}")

        if libur.lower() == 'y':
            print("Minggu ini dilewati (continue).")
            minggu += 1
            continue

        elif libur.lower() == 'n':
            pengeluaran = int(input("Masukkan pengeluaran minggu ini: "))
            print(f"Masukkan pengeluaran minggu ini: {pengeluaran}")
            saldo -= pengeluaran
            
            if saldo < 0:
                break

        minggu += 1
 
    else:
        print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}\n")
        continue
     
    break

else:
    print(f"Nobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")
```
