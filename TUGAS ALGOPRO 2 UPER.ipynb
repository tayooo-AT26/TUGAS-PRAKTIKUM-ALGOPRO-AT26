```python
#MENYIMPAN DATA DIRI GIANT DALAM SUATU VARIABEL

nama = 'Giant'
usia = 20            # tahun
berat_badan = 97.5   # kg 
tinggi_badan = 1.75  # meter 
status_olahraga = 'False'
print ('Saya bernama :', nama,)
print ('Saya berusia :', usia, 'tahun')
print ('Berat badan saya :', berat_badan, 'kg')
print ('Tinggi badan saya :', tinggi_badan, 'cm')
print ('Status olahraga :', status_olahraga)
```

```python
#GIANT MENCATAT BERAT BADAN SEBAGAI STRING '97.5'. KONVERSIKAN MENJADI FLOAT.

nama = 'Giant'
usia = 20              # tahun 
berat_string = "97.5"  # kg
berat = float(berat_string)
print(f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat} kg")
```

```python
#MENGHITUNG BMI GIANT 

berat = 63       # kg 
tinggi = 1.75    # meter
BMI = berat / (tinggi ** 2)
print("BMI asli :", BMI)
print("BMI 2 desimal :", round(BMI, 2))
```

```python
#TERGET MINUM AIR 

target_air = 2000       # ml
isi_botol = 350         # ml/botol

# True division
hasil_asli = target_air / isi_botol

# Floor division untuk jumlah botol penuh
botol_penuh = target_air // isi_botol

# Modulo untuk sisa air
sisa_air = target_air % isi_botol

print("\nHasil pembagian asli :", hasil_asli)
print("Botol penuh :", botol_penuh)
print("Sisa air :", sisa_air, "ml")
```
