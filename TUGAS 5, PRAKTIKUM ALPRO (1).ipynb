```python
hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input(f"Masukkan pengeluaran hari ke-{hari}: "))
    saldo -= pengeluaran
   
    if saldo <= 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
    
    else:
        print(f"Pengeluaran hari ke-{hari}: {pengeluaran}")
        print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")
        hari += 1
```

```python
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    minggu = i + 1
   
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{minggu}: Nobita tidak jajan, dilewati analisisnya.")
        continue
    
    print(f"Minggu ke-{minggu}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")

else:
    print("Analisis mingguan selesai.")
```

```python
for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")
    total_bulanan = 0
    berhenti = False

    for minggu in range(1, 5):
        pengeluaran = int(input(f"Pengeluaran minggu ke-{minggu}: "))
        total_bulanan += pengeluaran

        if pengeluaran > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            berhenti = True
            break

    if not berhenti:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")
```

```python
saldo_awal = 300000
saldo = saldo_awal
 
for bulan in range(11, 13):
    print(f"\n*** Bulan {bulan} ***")
    minggu = 1
 
    while minggu <= 4:
        libur = input(f"  Minggu {minggu} libur? (y/n) : ").strip().lower()
 
        if libur == "y":
            print("  >> Libur, tidak ada pengeluaran.")
            minggu += 1
            continue
        elif libur != "n":
            print("  [!] Jawab dengan 'y' atau 'n' saja.")
            continue  # ulangi minggu yang sama
 
        uang_keluar = int(input("  Pengeluaran minggu ini    : Rp"))
        saldo -= uang_keluar
 
        if saldo < 0:
            print(f"  >> Saldo minus Rp{abs(saldo)}! Simulasi dihentikan.")
            break
        minggu += 1
        
    else:
        print(f"  >> Saldo akhir bulan {bulan}: Rp{saldo}")
 
    if saldo < 0:
        break
else:
    print(f"\nNobita berhasil menabung {saldo} dari total {saldo_awal}")
 
```
