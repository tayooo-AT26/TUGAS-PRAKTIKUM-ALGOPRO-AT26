```python
#1. MEMBUAT DATA BELANJAAN 

#Buat sebuah list belanjaan 
list_barang = ['Apel', 'Susu', 'Roti']

#Harga dan jumlah barng 
harga_jumlah = [5000, 3, 12000, 2, 8000, 1]

#Tampilkan karakter 
print (list_barang [0][0])         #karakter pertama dari barang pertama, [0] dari paling kiri
print (list_barang [-1][-1])       #karakter terakhir dari barang terakhir, [-1] dari paling kanan
```

```python
#2. HITUNG TOTAL HARGA PER BARANG 

#Hitung total harga masing-masing barang 
harga_apel = harga_jumlah [0] * harga_jumlah[1]
harga_susu = harga_jumlah [2] * harga_jumlah[3]
harga_roti = harga_jumlah [4] * harga_jumlah[5]
print ('Harga keseluruhan Apel : Rp.', harga_apel)
print ('Harga keseluruhan Susu : Rp.', harga_susu)
print ('Harga keseluruhan Roti : Rp.', harga_roti)
```

```python
#2. HITUNG TOTAL HARGA PER BARANG 

#Hitung total keseluruhan belanja
harga_apel = 15000 
harga_susu = 24000
harga_roti = 8000
total_keseluruhan_belanja = harga_apel + harga_susu + harga_roti 
print ('Total keseluruhan belanja : Rp.', total_keseluruhan_belanja) 
```

```python
#3. CETAK STRUK BELANJA 

nama_pembeli = 'Nobita Nobi'
nama_toko = 'Toko Sembako Doraemon'
list_belanjaan = ['Apel', 'Susu', 'Roti']
harga_jumlah = [5000, 3, 12000, 2, 8000, 1]

#Struk belanja
print ('========================================\n'          #\n => untuk baris baru, menunjukkan akhir dari satu baris
    f'\t{nama_toko}\n'                                       #\t => memberikan indentasi atau tab
    f'========================================\n'
    f'Pembeli : {nama_pembeli}\n')

print (f'Barang  \tHarga \tJumlah \tTotal\n'
    f'----------------------------------------\n'
    f'{list_barang[0]}\t\t{harga_jumlah[0]} \t{harga_jumlah[1]} \t{harga_apel}\n'    #\t 2x => memberikan tab 2x, agar ke tengah  
    f'{list_barang[1]}\t\t{harga_jumlah[2]} \t{harga_jumlah[3]} \t{harga_susu}\n'
    f'{list_barang[2]}\t\t{harga_jumlah[4]} \t{harga_jumlah[5]} \t{harga_roti}\n')

print ('----------------------------------------\n'
    f'Total Belanja \t\t\t{total_keseluruhan_belanja}\n'                             #\t 3x => memberikan tab 3x, agar ke ujung kanan  
    f'========================================\n'
    f'    Terima kasih sudah berbelanja!👍\n'                                       #memunculkan emoji dengan (windows) plus (.)
    f'========================================\n')
       
```
